This notebook explores GeoAI tools for solar soiling detection:
- Zero-shot segmentation with SAM for solar array masks
- STAC searches with geo_agents for additional imagery
- Labeling assistance and data discovery

## 1. Setting Up the Development Environment

Install GeoAI and import necessary libraries.

```python
# Install GeoAI if not already
# !pip install geoai-py

# Imports
try:
    import geoai.sam as sam  # Segment Anything Model
    import geoai.geo_agents as ga  # Geo agents for STAC
    import geoai.map_tools as mt  # For visualization
    GEOAI_AVAILABLE = True
    print("GeoAI imported successfully")
except ImportError as e:
    print(f"GeoAI not available: {e}")
    GEOAI_AVAILABLE = False

import rasterio
import geopandas as gpd
from pathlib import Path
```

## 2. Defining the Problem and Requirements

We need to assist with labeling solar arrays in NAIP tiles and discover additional imagery.

Requirements:
- Zero-shot segmentation for solar panels using SAM
- STAC search for more NAIP/Planet data
- Visualize results interactively

## 3. Designing the Solution Architecture

- Load a sample NAIP tile
- Use SAM to generate masks for "solar panels"
- Use geo_agents to search STAC for additional data
- Visualize with map_tools

```python
# 4. Implementing Core Functionality

if GEOAI_AVAILABLE:
    # Load sample tile
    tile_path = Path("../data/tiles/tile_0_0.png")  # Adjust path
    if tile_path.exists():
        # SAM segmentation
        masks = sam.segment_image(str(tile_path), text_prompt="solar panels")
        print(f"Generated {len(masks)} masks")
        
        # STAC search
        results = ga.search_stac(collection="naip", bbox="-122.5,37.7,-122.3,37.9")  # Example bbox
        print(f"Found {len(results)} STAC items")
    else:
        print("Sample tile not found")
else:
    print("GeoAI not available")
```

```python
# 5. Adding Error Handling and Validation

try:
    if GEOAI_AVAILABLE:
        # Validate inputs
        if not tile_path.exists():
            raise FileNotFoundError("Tile file not found")
        # Add more validations
except Exception as e:
    print(f"Error: {e}")
```

```python
# 6. Testing the Implementation

# Simple test
if GEOAI_AVAILABLE:
    assert len(masks) >= 0, "Masks should be generated"
    assert len(results) >= 0, "STAC search should return items"
    print("Tests passed")
else:
    print("Skipping tests - GeoAI not available")
```
